Pattern 3 — CHAIN-OF-THOUGHT
Prompt

Explain Stack in Data Structures to a placement interview candidate. Think step by step before answering.

Step 1: define Stack in one sentence. Step 2: give one concrete example with code. Step 3: explain why interviewers ask about it. Step 4: synthesise into a 3-sentence summary.

Output
Step 1 — Definition

Stack in Data Structures describes how the execution time of an algorithm increases with input size.

Step 2 — Example
for i in range(n):
    print(i)

This loop runs n times, so the complexity is O(n).

for i in range(n):
    for j in range(n):
        print(i, j)

This runs n × n times, so the complexity is O(n²).

Step 3 — Why Interviewers Ask

Interviewers ask Stack questions to check whether students can write efficient and scalable code instead of only writing correct code.

Step 4 — Summary

Stack in Data Structures measures algorithm efficiency. It helps compare different solutions based on execution time and scalability. In interviews, understanding Stack shows strong problem-solving and optimization skills.
